### Business Understanding

#### What is the business problem?

Customer churn is a major challenge for telecom companies because acquiring a new customer is generally more expensive than retaining an existing one. The objective of this project is to predict whether a customer is likely to leave the company (churn) based on their demographic information, subscribed services, account details, and billing information.

By accurately identifying customers who are at high risk of churning, the company can implement targeted retention strategies such as personalized offers, discounts, loyalty programs, or proactive customer support.
